```python
import pandas as pd
import numpy as np
print("Libraries imported successfully!")
```

```python
# Loading the raw datasets
anime_df = pd.read_csv('data/anime.csv')
rating_df = pd.read_csv('data/rating.csv')

# Checking the initial size of data
print(f"Original Anime Data: {anime_df.shape}")
print(f"Original Rating Data: {rating_df.shape}")
```

```python
# 1. Cleaning the Anime Data (Droping rows with missing Name or Genre)
anime_df = anime_df.dropna(subset=['name', 'genre'])

# 2. Cleaning the Rating Data (Removing '-1' ratings which mean 'unrated')
rating_df = rating_df[rating_df['rating'] != -1]

# 3. Checking the size of data after cleaning
print(f"Cleaned Anime Data: {anime_df.shape}")
print(f"Cleaned Rating Data: {rating_df.shape}")

# Showing the first 5 rows to verify
anime_df.head()
```

```python
# Save the cleaned data for Phase 2
anime_df.to_csv('data/cleaned_anime.csv', index=False)
rating_df.to_csv('data/cleaned_rating.csv', index=False)
print("Cleaned datasets saved successfully to the 'data' folder!")
```
